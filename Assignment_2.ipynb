```python
import pandas as pd
import numpy as np

# Load the dataset; make sure 'train.csv' is in the current working directory.
df = pd.read_csv('/train.csv')

# Display the first few rows to understand the dataset structure
print(df.head())
```

```python
print(df.columns)
```

```python
# Check missing values in each column
print("Missing values per column:")
print(df.isnull().sum())

# Impute missing values.
# Process numeric columns: fill missing with the median
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Process object (categorical) columns: fill missing with the mode
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

print("\nAfter imputation, missing values per column:")
print(df.isnull().sum())
```

```python
# Define a helper function to remove non-numeric characters (except the decimal point)
def extract_numeric(series):
    # This regex keeps digits and dots, and then convert to float.
    return series.astype(str).str.replace('[^0-9.]', '', regex=True).replace('', np.nan).astype(float)

# Remove units from specific columns:
# Adjust column names based on your CSV; typical names might be "Mileage", "Engine", "Power", "New_price".
df['Mileage'] = extract_numeric(df['Mileage'])
df['Engine'] = extract_numeric(df['Engine'])
df['Power'] = extract_numeric(df['Power'])
df['New_price'] = extract_numeric(df['New_Price'])

# Verify the changes
print(df[['Mileage', 'Engine', 'Power', 'New_price']].head())
```

```python
# Check the unique categories
print("Unique values in Fuel_Type:", df['Fuel_Type'].unique())
print("Unique values in Transmission:", df['Transmission'].unique())

# One-hot encode the categorical variables
df = pd.get_dummies(df, columns=['Fuel_Type', 'Transmission'], drop_first=True)

# Verify the transformation by looking at the new columns
print(df.head())
```

```python
# Get the current year; using pd.Timestamp.now() ensures that it is dynamic
current_year = pd.Timestamp.now().year

# Create a new column 'Age' by subtracting the manufacture year from the current year
df['Age'] = current_year - df['Year']

# Display the updated dataframe to see the new "Age" column
print(df[['Year', 'Age']].head())
```

```python
import pandas as pd
import numpy as np

separator = "\n" + "=" * 60 + "\n"

# ------------------ STEP 1: Create 'Age' from 'Year' ------------------
print(separator)
print("STEP 1: Create 'Age' from 'Year'")
if 'Year' in df.columns:
    current_year = pd.Timestamp.now().year
    df['Age'] = current_year - df['Year']
    print("New 'Age' column created (first 5 rows):")
    print(df[['Year', 'Age']].head())
else:
    print("Column 'Year' not found. Cannot compute 'Age'.")

# ------------------ STEP 2: Select Columns ------------------
print(separator)
print("STEP 2: Select Columns: Price, Mileage, Age, and Fuel_Type (if available)")
columns_to_select = ['Price', 'Mileage', 'Age']
if 'Fuel_Type' in df.columns:
    columns_to_select.append('Fuel_Type')
selected_df = df[columns_to_select]
print("Selected Columns (first 5 rows):")
print(selected_df.head())

# ------------------ STEP 3: Filter Rows ------------------
print(separator)
print("STEP 3: Filter Rows (Keep only cars with Price > 3)")
filtered_df = df[df['Price'] > 3]
print("Filtered Data (first 5 rows):")
print(filtered_df.head())

# ------------------ STEP 4: Rename a Column ------------------
print(separator)
print("STEP 4: Rename 'Mileage' to 'Mileage_kmpl'")
df_renamed = df.rename(columns={'Mileage': 'Mileage_kmpl'})
print("Data with Renamed Column (first 5 rows):")
print(df_renamed.head())

# ------------------ STEP 5: Create a New Column (Mutate) ------------------
print(separator)
print("STEP 5: Create New Column 'Cost_Efficiency' = Price / Mileage")
df['Cost_Efficiency'] = np.where(df['Mileage'] > 0, df['Price'] / df['Mileage'], np.nan)
print("Data with New Column 'Cost_Efficiency' (first 5 rows):")
print(df[['Price', 'Mileage', 'Cost_Efficiency']].head())

# ------------------ STEP 6: Arrange (Sort Data) ------------------
print(separator)
print("STEP 6: Sort Data by Price in Descending Order")
df_sorted = df.sort_values(by='Price', ascending=False)
print("Data Sorted by Price (first 5 rows):")
print(df_sorted[['Price', 'Mileage', 'Age']].head())

# ------------------ STEP 7: Group By and Summarize ------------------
print(separator)
print("STEP 7: Group By (Summary) by Owner_Type (if available)")
if 'Owner_Type' in df.columns:
    summary = df.groupby('Owner_Type').agg(
        Avg_Price=('Price', 'mean'),
        Count=('Price', 'count'),
        Avg_Age=('Age', 'mean'),
        Avg_Mileage=('Mileage', 'mean')
    ).reset_index()
    print("Summary Grouped by Owner_Type:")
    print(summary)
else:
    print("Column 'Owner_Type' not found for grouping. Consider using another categorical variable for grouping.")

print(separator)
```
